# End-to-End Health Data Analysis: Heart Disease Prediction
## Domain: Healthcare Analytics

### Project Objective
Analyze patient clinical parameters from the `HeartDiseaseTrain-Test.csv` dataset and build a machine learning classification model to predict whether a patient has heart disease.

### Dataset Features:
- `age`: Age in years
- `sex`: Male or Female
- `chest_pain_type`: Type of chest pain (e.g., Typical angina, Asymptomatic)
- `resting_blood_pressure`: Resting blood pressure (in mm Hg)
- `cholestoral`: Serum cholesterol in mg/dl
- `fasting_blood_sugar`: Lower or Greater than 120 mg/ml
- `rest_ecg`: Resting electrocardiographic results
- `Max_heart_rate`: Maximum heart rate achieved
- `exercise_induced_angina`: Yes or No
- `oldpeak`: ST depression induced by exercise relative to rest
- `slope`: The slope of the peak exercise ST segment
- `vessels_colored_by_flourosopy`: Number of major vessels colored
- `thalassemia`: Thalassemia defect type
- `target`: 1 = Disease, 0 = No Disease

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Set visualization styles
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load the user's specific dataset
df = pd.read_csv("HeartDiseaseTrain-Test.csv")

# Display the first 5 rows
print("Dataset Loaded Successfully. Shape:", df.shape)
df.head()
```

## 1. Exploratory Data Analysis (EDA)
We will explore the dataset to understand the distributions of features, check for missing values, and investigate how different clinical attributes relate to the target variable.

```python
# Check for missing values and data types
print("--- Data Info ---")
print(df.info())

print("\n--- Missing Values Summary ---")
print(df.isnull().sum())

print("\n--- Statistical Summary (Numerical Features) ---")
display(df.describe())
```

### Graph 1: Target Variable Distribution
Let's look at the balance of our target classes to ensure the dataset isn't heavily skewed toward one outcome.

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='target', data=df, palette='Set2')
plt.title('Distribution of Heart Disease Cases (0 = No, 1 = Yes)')
plt.xlabel('Target Outcome')
plt.ylabel('Patient Count')
plt.show()
```

### Graph 2: Correlation Matrix Heatmap
We compute the Pearson correlation matrix to see how the numerical clinical metrics scale or decline alongside each other. *(Note: Categorical text columns are excluded from this specific plot).*

```python
plt.figure(figsize=(10, 8))
# Select only numeric columns for the initial correlation matrix
numeric_cols = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_cols.corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Numerical Clinical Features')
plt.show()
```

### Graph 3: Age Distribution Segmented by Target
We will analyze how age distributes across patients diagnosed with and without heart disease using a Kernel Density Estimate (KDE) plot.

```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x='age', hue='target', fill=True, common_norm=False, palette='crest', alpha=0.5)
plt.title('Patient Age Distribution by Heart Disease Status')
plt.xlabel('Age (Years)')
plt.ylabel('Density')
plt.show()
```

### Graph 4: Cardiovascular Efficiency (Max Heart Rate vs Age)
Let's see if there is a visible boundary when comparing a patient's maximum achieved heart rate (`Max_heart_rate`) against their `age`, categorized by the target label.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='age', y='Max_heart_rate', hue='target', data=df, palette='bwr', alpha=0.8)
plt.title('Max Heart Rate Achieved vs Age')
plt.xlabel('Age')
plt.ylabel('Maximum Heart Rate')
plt.legend(title='Target (0=Safe, 1=Disease)')
plt.show()
```

## 2. Data Preprocessing (Encoding & Scaling)
Because this dataset contains text-based categorical variables (like "Male", "Typical angina"), we must convert them into a machine-readable numeric format using **One-Hot Encoding**. Then, we will scale our numerical features so that larger values (like cholesterol) do not overpower smaller values (like oldpeak).

```python
# Separate target from features
X = df.drop('target', axis=1)
y = df['target']

# 1. One-Hot Encoding for categorical (text) columns
# drop_first=True helps avoid the dummy variable trap
X_encoded = pd.get_dummies(X, drop_first=True)

# 2. Split into 80% training and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42, stratify=y)

# 3. Standardize all features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Original Feature Count:", X.shape[1])
print("Encoded Feature Count:", X_encoded.shape[1])
print("Training Data Shape:", X_train_scaled.shape)
```

## 3. Model Training & Evaluation
We will train a **Random Forest Classifier** on our scaled and encoded dataset. We'll evaluate its performance using a confusion matrix.

```python
# Initialize and train the Random Forest Classifier
model = RandomForestClassifier(random_state=42)
model.fit(X_train_scaled, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test_scaled)

# Evaluate the model
print("Classification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['No Disease', 'Disease'], yticklabels=['No Disease', 'Disease'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

print("\nAccuracy Score:", accuracy_score(y_test, y_pred))
```

### Graph 5: Top 10 Feature Importances
Since One-Hot Encoding expanded our feature count, we will look at the top 10 most influential features driving our model's decisions.

```python
importances = model.feature_importances_
feature_names = X_encoded.columns
indices = np.argsort(importances)[::-1]

# Take only the top 10 features for a cleaner graph
top_n = 10
top_indices = indices[:top_n]

plt.figure(figsize=(10, 6))
sns.barplot(x=importances[top_indices], y=feature_names[top_indices], hue=feature_names[top_indices], palette='viridis', legend=False)
plt.title('Top 10 Feature Importances in Heart Disease Prediction')
plt.xlabel('Relative Importance Score')
plt.ylabel('Clinical Features')
plt.show()
```

## 4. Key Findings & Conclusions
- **Preprocessing Impact:** One-Hot Encoding successfully translated clinical text records into a viable machine learning format.
- **Primary Predictors:** After evaluating feature importance, attributes like `Max_heart_rate`, `age`, and specific forms of `chest_pain_type` strongly influence the classification.
- **Model Performance:** The Random Forest Classifier provides a robust baseline. By reviewing the Confusion Matrix, we can identify how many False Negatives occurred (which are critical to minimize in healthcare diagnostics).
